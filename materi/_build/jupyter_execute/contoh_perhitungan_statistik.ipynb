# Contoh Perhitungan Statistik – PSD

Notebook ini memverifikasi contoh perhitungan statistik yang digunakan dalam `statistik_knime.md`. Contoh data: `2, 4, 4, 6, 8`.

```python
import math
import statistics

data = [2, 4, 4, 6, 8]
n = len(data)
mean = sum(data) / n
median = statistics.median(data)
minimum = min(data)
maximum = max(data)
sample_variance = statistics.variance(data)
sample_std = statistics.stdev(data)

print('Data:', data)
print('n =', n)
print('Sum =', sum(data))
print('Mean =', mean)
print('Median =', median)
print('Minimum =', minimum)
print('Maximum =', maximum)
print('Sample variance =', sample_variance)
print('Sample standard deviation =', sample_std)
```

## Perhitungan Mean

$$\bar{x}=\frac{\sum x_i}{n}$$

Untuk `2, 4, 4, 6, 8`:

$$\bar{x}=\frac{2+4+4+6+8}{5}=4.8$$

```python
total = sum(data)
manual_mean = total / n
print(f'Sum = {total}')
print(f'Mean = {manual_mean}')
```

## Variance dan Standard Deviation

$$s^2=\frac{\sum(x_i-\bar{x})^2}{n-1}$$

$$s=\sqrt{s^2}$$

```python
squared_deviations = [(x - mean) ** 2 for x in data]
sum_squared_deviations = sum(squared_deviations)
manual_variance = sum_squared_deviations / (n - 1)
manual_std = math.sqrt(manual_variance)

print('Squared deviations:', squared_deviations)
print('Sum squared deviations =', sum_squared_deviations)
print('Sample variance =', manual_variance)
print('Sample standard deviation =', manual_std)
```

## Min, Median, Max

```python
print('Minimum =', min(data))
print('Median =', statistics.median(data))
print('Maximum =', max(data))
```
